```python
import matplotlib.pyplot as plt
import numpy as np

# Datos de la tabla
categorias = ['<2ha', '2-4 ha', '4-10ha', '>10ha']
barlovento = [76.47, 8.99, 7.10, 7.43]
slm = [46.10, 21.42, 18.01, 14.47]

# Configuración del gráfico
fig, ax = plt.subplots(figsize=(10, 6))

# Posiciones de las barras
x = np.arange(len(categorias))
ancho = 0.35

# Crear barras
barras1 = ax.bar(x - ancho/2, barlovento, ancho,
                label='Barlovento', color='#404040', edgecolor='black')
barras2 = ax.bar(x + ancho/2, slm, ancho,
                label='SLM', color='#808080', edgecolor='black')

# Personalización del gráfico
ax.set_ylabel('Porcentaje (%)', fontname='Arial', fontsize=12)
ax.set_xlabel('Categoría de Tamaño de Parcela', fontname='Arial', fontsize=12)
ax.set_title('Distribución Regional de Tamaño de Parcelas',
             fontname='Arial', fontsize=14, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(categorias, fontname='Arial')
ax.set_ylim(0, 85)
ax.legend(loc='upper right', frameon=True, fontsize=10)

# Añadir valores encima de las barras
def añadir_etiquetas(barras):
    for barra in barras:
        altura = barra.get_height()
        ax.text(barra.get_x() + barra.get_width()/2., altura + 1,
                f'{altura:.2f}%', ha='center', va='bottom',
                fontname='Arial', fontsize=9)

añadir_etiquetas(barras1)
añadir_etiquetas(barras2)

# Añadir cuadrícula
ax.grid(axis='y', alpha=0.3, linestyle='--')
ax.set_axisbelow(True)

# Ajustar diseño
plt.tight_layout()

# Mostrar gráfico
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from typing import Optional, Tuple

def configurar_estilo_grafico():
    """
    Configura el estilo global de matplotlib para gráficos limpios y minimalistas.
    """
    plt.style.use('default')

    plt.rcParams.update({
        'figure.figsize': (10, 6),
        'figure.dpi': 100,
        'savefig.dpi': 300,
        'font.size': 12,
        'axes.titlesize': 16,
        'axes.labelsize': 14,
        'xtick.labelsize': 12,
        'ytick.labelsize': 12,
        'axes.linewidth': 1.2,
        'axes.spines.top': False,
        'axes.spines.right': False,
        'axes.grid': False,
        'font.family': 'sans-serif',
        'font.sans-serif': ['Arial', 'DejaVu Sans', 'Liberation Sans']
    })

def crear_grafico_barras_parcelas():
    """
    Crea un gráfico de barras minimalista para los datos de parcelas agrícolas
    adaptado al nuevo estilo visual.
    """
    # Configurar estilo
    configurar_estilo_grafico()

    # Datos originales de la tabla
    categorias = ['<2ha', '2-4 ha', '4-10ha', '>10ha']

    # Conteos reales (para las alturas de las barras)
    barlovento_conteos = [689, 81, 64, 67]
    slm_conteos = [325, 151, 127, 102]

    # Crear figura y ejes
    fig, ax = plt.subplots(figsize=(10, 6))

    # Posiciones de las barras
    x = np.arange(len(categorias))
    ancho_barras = 0.6

    # Colores en escala de grises
    color_barlovento = '#404040'
    color_slm = '#808080'
    color_puntos = '#A0A0A0'

    # 1. DIBUJAR BARRAS PRINCIPALES
    barras_barlovento = ax.bar(x - ancho_barras/2, barlovento_conteos, ancho_barras/2,
                              color=color_barlovento, alpha=0.8, edgecolor='black', linewidth=1)

    barras_slm = ax.bar(x + ancho_barras/2, slm_conteos, ancho_barras/2,
                       color=color_slm, alpha=0.8, edgecolor='black', linewidth=1)

    # 2. CALCULAR Y DIBUJAR DESVIACIONES ESTÁNDAR (simuladas para este ejemplo)
    # Como tenemos conteos, simulamos una desviación estándar basada en los datos
    barlovento_std = [val * 0.15 for val in barlovento_conteos]  # 15% del valor
    slm_std = [val * 0.15 for val in slm_conteos]  # 15% del valor

    # Añadir líneas de desviación estándar (solo hacia arriba)
    for i, (pos, promedio, desv_std) in enumerate(zip(x - ancho_barras/2, barlovento_conteos, barlovento_std)):
        # Línea vertical de desviación estándar
        ax.plot([pos, pos], [promedio, promedio + desv_std],
               color='black', linewidth=2, alpha=0.8)

        # Línea horizontal superior (reducida en 70%)
        ancho_linea_horizontal = (ancho_barras/2) * 0.075
        ax.plot([pos - ancho_linea_horizontal, pos + ancho_linea_horizontal],
               [promedio + desv_std, promedio + desv_std],
               color='black', linewidth=2, alpha=0.8)

    for i, (pos, promedio, desv_std) in enumerate(zip(x + ancho_barras/2, slm_conteos, slm_std)):
        # Línea vertical de desviación estándar
        ax.plot([pos, pos], [promedio, promedio + desv_std],
               color='black', linewidth=2, alpha=0.8)

        # Línea horizontal superior (reducida en 70%)
        ancho_linea_horizontal = (ancho_barras/2) * 0.075
        ax.plot([pos - ancho_linea_horizontal, pos + ancho_linea_horizontal],
               [promedio + desv_std, promedio + desv_std],
               color='black', linewidth=2, alpha=0.8)

    # 3. AÑADIR PUNTOS INDIVIDUALES CON TRANSPARENCIA (simulados)
    # Generamos puntos aleatorios alrededor de los valores reales
    np.random.seed(42)

    for i, categoria in enumerate(categorias):
        # Puntos para Barlovento
        n_puntos_barlovento = max(5, barlovento_conteos[i] // 100)  # Número reducido de puntos para visualización
        y_jitter_barlovento = np.random.normal(barlovento_conteos[i], barlovento_std[i] * 0.3, n_puntos_barlovento)
        x_jitter_barlovento = np.random.normal(x[i] - ancho_barras/2, 0.02, n_puntos_barlovento)

        # Puntos para SLM
        n_puntos_slm = max(5, slm_conteos[i] // 100)
        y_jitter_slm = np.random.normal(slm_conteos[i], slm_std[i] * 0.3, n_puntos_slm)
        x_jitter_slm = np.random.normal(x[i] + ancho_barras/2, 0.02, n_puntos_slm)

        # Dibujar puntos
        ax.scatter(x_jitter_barlovento, y_jitter_barlovento,
                  color=color_puntos, alpha=0.5, s=20, edgecolors='none')
        ax.scatter(x_jitter_slm, y_jitter_slm,
                  color=color_puntos, alpha=0.5, s=20, edgecolors='none')

    # 4. CONFIGURAR EJES Y ETIQUETAS
    ax.set_ylabel('N° Parcelas', fontweight='bold')
    ax.set_xlabel('Categorías de Tamaño', fontweight='bold')
    ax.set_title('Distribución de Parcelas por Categoría de Tamaño',
                fontweight='bold', pad=20)

    # Configurar etiquetas del eje X
    ax.set_xticks(x)
    ax.set_xticklabels(categorias, rotation=0, ha='center')

    # Ajustar límites del eje Y
    y_max = max(max(barlovento_conteos), max(slm_conteos)) * 1.2
    ax.set_ylim(0, y_max)

    # 5. CONFIGURAR APARIENCIA DE EJES (solo líneas X e Y)
    ax.spines['left'].set_linewidth(1.2)
    ax.spines['bottom'].set_linewidth(1.2)

    # Configurar ticks
    ax.tick_params(axis='both', which='major', labelsize=12, width=1.2)
    ax.tick_params(axis='x', which='major', length=6)
    ax.tick_params(axis='y', which='major', length=6)

    # 6. NO AÑADIR LEYENDA (eliminada completamente)

    # 7. AJUSTAR LAYOUT
    plt.tight_layout()

    return fig

def crear_grafico_alternativo():
    """
    Versión alternativa con un solo grupo de barras para comparación.
    """
    configurar_estilo_grafico()

    categorias = ['<2ha', '2-4 ha', '4-10ha', '>10ha']
    barlovento_conteos = [689, 81, 64, 67]

    fig, ax = plt.subplots(figsize=(8, 6))

    x = np.arange(len(categorias))
    ancho_barras = 0.6

    # Barras principales
    barras = ax.bar(x, barlovento_conteos, ancho_barras,
                   color='#666666', alpha=0.8, edgecolor='black', linewidth=1)

    # Desviaciones estándar simuladas
    std_values = [val * 0.12 for val in barlovento_conteos]

    # Líneas de desviación estándar
    for i, (pos, promedio, desv_std) in enumerate(zip(x, barlovento_conteos, std_values)):
        ax.plot([pos, pos], [promedio, promedio + desv_std],
               color='black', linewidth=2, alpha=0.8)

        ancho_linea_horizontal = ancho_barras * 0.075
        ax.plot([pos - ancho_linea_horizontal, pos + ancho_linea_horizontal],
               [promedio + desv_std, promedio + desv_std],
               color='black', linewidth=2, alpha=0.8)

    # Puntos individuales
    np.random.seed(42)
    for i, categoria in enumerate(categorias):
        n_puntos = max(8, barlovento_conteos[i] // 80)
        y_jitter = np.random.normal(barlovento_conteos[i], std_values[i] * 0.4, n_puntos)
        x_jitter = np.random.normal(x[i], 0.03, n_puntos)

        ax.scatter(x_jitter, y_jitter,
                  color='#999999', alpha=0.5, s=25, edgecolors='none')

    # Configuración final
    ax.set_ylabel('N° Parcelas', fontweight='bold')
    ax.set_xlabel('Categorías de Tamaño', fontweight='bold')
    ax.set_title('Distribución de Parcelas - Barlovento', fontweight='bold', pad=20)
    ax.set_xticks(x)
    ax.set_xticklabels(categorias)
    ax.set_ylim(0, max(barlovento_conteos) * 1.25)

    plt.tight_layout()
    return fig

# EJECUCIÓN PRINCIPAL
if __name__ == "__main__":
    print("Generando gráficos adaptados al nuevo estilo visual...")

    # Gráfico principal con dos grupos
    fig1 = crear_grafico_barras_parcelas()
    plt.savefig('grafico_parcelas_adaptado.png', dpi=300, bbox_inches='tight')
    print("✅ Gráfico principal guardado como 'grafico_parcelas_adaptado.png'")

    # Gráfico alternativo
    fig2 = crear_grafico_alternativo()
    plt.savefig('grafico_alternativo.png', dpi=300, bbox_inches='tight')
    print("✅ Gráfico alternativo guardado como 'grafico_alternativo.png'")

    plt.show()

    print("\n" + "="*60)
    print("ADAPTACIÓN COMPLETADA EXITOSAMENTE")
    print("="*60)
    print("Características implementadas:")
    print("✓ Paleta de colores en escala de grises")
    print("✓ Barras que muestran valores promedio (conteos)")
    print("✓ Líneas de desviación estándar (simuladas, solo hacia arriba)")
    print("✓ Puntos individuales con transparencia del 50%")
    print("✓ Solo líneas en los ejes X e Y (sin cuadrícula)")
    print("✓ Sin valores numéricos visibles en el gráfico")
    print("✓ Sin leyenda")
    print("✓ Línea horizontal de desviación estándar reducida en 70%")
    print("✓ CON etiquetas de categorías en el eje X")
```
