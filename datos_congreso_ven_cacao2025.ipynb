Extract unique and duplicate rows from the file "/content/no_depurado_censo_slm-bar-barlov_datos_survey2025_.xlsx" based on the combination of the "Nombre_Pro" and "CI_RIF_Pro" columns and save them into separate Excel files.

## Load the data

### Subtask:
Load the data from the specified Excel file into a pandas DataFrame.


**Reasoning**:
Load the data from the specified Excel file into a pandas DataFrame and display its head and info.



```python
import pandas as pd

file_path = "/content/no_depurado_censo_slm-bar-barlov_datos_survey2025_.xlsx"

df = pd.read_excel(file_path)

# Convert specified columns to float, coercing errors to NaN
float_cols = ["Sup_UP_ha", "Sup_Cacao_", "Sup_ha_calc_geom", "Shape_Area_calc_geom", "lat_dec", "long_dec"]
for col in float_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')

display(df.head())
display(df.info())
```

## Identify unique and duplicate rows

### Subtask:
Based on the combination of the "Nombre_Pro" and "CI_RIF_Pro" columns, identify the unique and duplicate rows.


**Reasoning**:
Create boolean masks to identify duplicate and unique rows based on the specified columns and then use these masks to create new DataFrames for duplicate and unique rows respectively.



```python
duplicate_mask = df.duplicated(subset=['Nombre_Pro', 'CI_RIF_Pro'], keep=False)
df_duplicates = df[duplicate_mask].copy()
df_unique = df[~duplicate_mask].copy()

print("Shape of original DataFrame:", df.shape)
print("Shape of duplicate rows DataFrame:", df_duplicates.shape)
print("Shape of unique rows DataFrame:", df_unique.shape)

display(df_duplicates.head())
display(df_unique.head())
```

## Save unique data

### Subtask:
Save the unique rows into a new Excel file.


**Reasoning**:
Save the unique rows DataFrame to an Excel file without including the index.



```python
unique_output_path = "/content/unique_rows.xlsx"
df_unique.to_excel(unique_output_path, index=False)
```

## Save duplicate data

### Subtask:
Save the duplicate rows into another new Excel file.


**Reasoning**:
Save the duplicate rows into a new Excel file.



```python
duplicate_output_path = "/content/duplicate_rows.xlsx"
df_duplicates.to_excel(duplicate_output_path, index=False)
```

## Split variety strings and clean names

```python
# Split the 'Variedad de cacao' column by comma and explode the list of varieties
df_varieties = df_unique['Variedad de cacao'].astype(str).str.split(',').explode().reset_index()

# Rename the columns
df_varieties.columns = ['original_index', 'individual_variety']

# Clean individual variety names (remove leading/trailing spaces, convert to lowercase)
df_varieties['individual_variety'] = df_varieties['individual_variety'].str.strip().str.lower()

# Merge with the original DataFrame to get the REGION column
df_varieties = pd.merge(df_varieties, df_unique[['REGION']], left_on='original_index', right_index=True)


display(df_varieties.head())
```

## Calculate frequency of individual varieties by region

### Subtask:
Group by region and individual variety and count occurrences.

Group the DataFrame containing individual varieties by 'REGION' and 'individual_variety' and count the occurrences of each individual variety in each region.

```python
individual_variety_counts = df_varieties.groupby(['REGION', 'individual_variety']).size().reset_index(name='count')
display(individual_variety_counts.head())
```

## Determine top individual varieties per region

### Subtask:
For each region, identify the top N most frequent individual cacao varieties.

**Reasoning**:
For each region, sort the individual varieties by their counts in descending order and select the top N varieties. I will use N=5 to identify the top 5 individual varieties per region.

```python
top_n_individual = 5
top_individual_varieties_by_region = individual_variety_counts.groupby('REGION').apply(lambda x: x.nlargest(top_n_individual, 'count')).reset_index(drop=True)
display(top_individual_varieties_by_region)
```

## Visualize frequency of top individual varieties

### Subtask:

## Agronomist Analysis of Fertilizer Use by Region and Parcel Size (Excluding Barinas)

Observing the distribution of fertilizer types across the 'Barlovento' and 'Sur Lago Maracaibo' regions, segmented by parcel size category, several key insights emerge for a data-driven agronomist:

**Overall Trends and Regional Differences:**

*   **Dominance of 'No aplica' and 'No sabe':** A significant portion of respondents in both 'Barlovento' and 'Sur Lago Maracaibo' indicated 'No aplica' or 'No sabe' when asked about fertilizer types. This highlights a critical data gap. From an agronomic standpoint, understanding fertilizer practices is fundamental to optimizing yield and soil health. The high frequency of these responses suggests either a lack of consistent fertilization practices, inadequate knowledge about the fertilizers being used, or limitations in data collection. This needs further investigation in the field to determine the underlying reasons.
*   **Regional Differences in Reported Practices:** While 'No aplica'/'No sabe' are prevalent in both regions, there are discernible differences in the reported use of specific fertilizer types when those categories are excluded.
    *   'Sur Lago Maracaibo' shows a more diverse reported use of fertilizer types, including 'Abono organico', 'Cal Agricola', 'Fertilizante foliar', and 'Fertilizante granulado'. This suggests a potentially broader adoption of different fertilization strategies in this region.
    *   'Barlovento', excluding the 'No sabe'/'No aplica' responses, still has a considerable number of entries that might require further clarification or indicate less standardized practices compared to 'Sur Lago Maracaibo'.

**Influence of Parcel Size:**

*   **Smaller Parcels (<2ha):** In both regions, smaller parcels show a notable frequency of reported fertilizer use (excluding 'No sabe'/'No aplica'). This could indicate that smaller-scale farmers are actively applying fertilizers, possibly due to more intensive management practices on limited land or easier access to inputs for smaller areas.
