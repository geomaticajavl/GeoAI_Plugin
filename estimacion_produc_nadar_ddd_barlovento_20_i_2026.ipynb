## Cargar Datos

### Subtask:
Cargar el archivo CSV 'estimacion_produc_nadar_ddd_barlovento_20-i-2026.csv' en un DataFrame de pandas.


**Reasoning**:
To load the CSV file into a pandas DataFrame and display its first rows, I need to import pandas, read the CSV, and then use the `head()` method.



```python
import pandas as pd

df = pd.read_csv('/content/estimacion_produc_nadar_ddd_barlovento_20-i-2026.csv')
print("DataFrame loaded successfully. Displaying the first 5 rows:")
df.head()
```

## Simulación de Monte Carlo y Actualización

### Subtask:
Para cada fila del DataFrame, realizar una simulación de Monte Carlo para generar nuevos valores para 'fact_prod_ML' basándose en 'Area (ha)'.


**Reasoning**:
To perform the Monte Carlo simulation and update the 'fact_prod_ML' column, I need to define the simulation parameters (iterations, range), then iterate through each row of the DataFrame, generate random numbers, calculate their mean, and assign it to the specified column.



```python
import numpy as np

# 1. Define el número de iteraciones para la simulación de Monte Carlo
num_iterations = 7500

# 2. Define el rango para la simulación del área
lower_bound = 0.770
upper_bound = 0.935

# 3. Itera sobre cada fila del DataFrame y realiza la simulación
for index, row in df.iterrows():
    # a. Genera una lista de números aleatorios distribuidos uniformemente
    simulated_values = np.random.uniform(lower_bound, upper_bound, num_iterations)

    # b. Calcula la media de estos valores simulados
    mean_simulated_value = np.mean(simulated_values)

    # c. Asigna esta media a la columna 'fact_prod_ML' de la fila actual
    df.loc[index, 'fact_prod_ML'] = mean_simulated_value

print("Monte Carlo simulation completed and 'fact_prod_ML' updated. Displaying the first 5 rows with updated values:")
df.head()
```

```python
# Importa la librería numpy, fundamental para operaciones numéricas, especialmente para la generación de números aleatorios en simulaciones de Monte Carlo.
import numpy as np

# Esta sección aplica métodos estocásticos, específicamente la simulación de Monte Carlo,
# para estimar y asignar valores a columnas clave del DataFrame. Este enfoque permite
# incorporar la incertidumbre y variabilidad inherente en los procesos de producción
# agrícola, mejorando la robustez de las estimaciones para la trazabilidad y la planificación.

# --- Simulación de Monte Carlo para 'fact_prod_ML' ---
# Objetivo: Estimar el 'factor de productividad de la planta' (fact_prod_ML) para cada parcela.
# Este factor es crucial para la estimación de la producción, reflejando la eficiencia intrínseca del cultivo.
# Se utiliza una simulación de Monte Carlo para reflejar la variabilidad esperada de este factor.

# 1. Define el número de iteraciones para la simulación de Monte Carlo.
# Un alto número de iteraciones (7500 en este caso) asegura que la media de los valores simulados
# sea una estimación robusta y estadísticamente significativa del valor esperado del factor, reduciendo el error de muestreo.
num_iterations = 7500

# 2. Define el rango (límites inferior y superior) para la simulación del 'fact_prod_ML'.
# Estos límites (0.770 a 0.935) deben basarse en datos históricos, estudios agronómicos o
# conocimiento experto del cultivo y la región, representando el rango de variabilidad plausible del factor.
lower_bound = 0.770
upper_bound = 0.935

# 3. Itera sobre cada fila del DataFrame para aplicar la simulación individualmente a cada parcela.
for index, row in df.iterrows():
    # a. Genera una lista de números aleatorios distribuidos uniformemente dentro del rango definido.
    # La distribución uniforme asume que cualquier valor dentro del rango es igualmente probable,
    # lo cual es una simplificación común en las primeras etapas de modelado de incertidumbre.
    simulated_values = np.random.uniform(lower_bound, upper_bound, num_iterations)

    # b. Calcula la media de estos valores simulados. La media es utilizada como el valor
    # representativo del 'fact_prod_ML' para la parcela actual, promediando la incertidumbre de la simulación.
    mean_simulated_value = np.mean(simulated_values)

    # c. Asigna esta media a la columna 'fact_prod_ML' de la fila actual.
    # Esto actualiza el DataFrame con el valor simulado para cada parcela.
    df.loc[index, 'fact_prod_ML'] = mean_simulated_value

# --- Simulación de Monte Carlo para 'fact_arrim' ---
# Objetivo: Estimar el 'factor de rendimiento del arrime' (fact_arrim), que representa la proporción
# de la producción cosechada que efectivamente se 'arrima' o es utilizable.
# Similar al 'fact_prod_ML', se simula para capturar su variabilidad.

# 1. Define el número de iteraciones para la simulación de Monte Carlo (reutilizamos num_iterations).
# El mismo número de iteraciones se mantiene para consistencia en la robustez de las simulaciones.

# 2. Define el rango para la simulación de 'fact_arrim'.
# Estos límites (0.7734 a 0.8513) reflejan el rango de eficiencia de arrime esperado, basado en
# factores como prácticas de cosecha, logística y condiciones del producto.
lower_bound_arrim = 0.7734
upper_bound_arrim = 0.8513

# 3. Itera sobre cada fila del DataFrame y realiza la simulación para 'fact_arrim'.
for index, row in df.iterrows():
    # a. Genera una lista de números aleatorios distribuidos uniformemente.
    simulated_values_arrim = np.random.uniform(lower_bound_arrim, upper_bound_arrim, num_iterations)

    # b. Calcula la media de estos valores simulados.
    mean_simulated_value_arrim = np.mean(simulated_values_arrim)

    # c. Asigna esta media a la columna 'fact_arrim' de la fila actual.
    df.loc[index, 'fact_arrim'] = mean_simulated_value_arrim

# --- Simulación de Monte Carlo para 'produc_simul_kg_haMC' ---
# Objetivo: Añadir una nueva columna con la producción simulada en kg por hectárea (produc_simul_kg_haMC).
# Esta simulación proporciona una estimación de la producción por unidad de área, incorporando la variabilidad.

# 1. Define el número de iteraciones para esta simulación de Monte Carlo.
# Un mayor número de iteraciones (15000) puede ser usado para variables con mayor incertidumbre o donde se requiere mayor precisión.
num_iterations_simul = 15000

# 2. Define el rango para la simulación de 'produc_simul_kg_haMC'.
# Los límites (326.4532 a 390.8709 kg/ha) representan el rango esperado de producción por hectárea,
# obtenido de datos históricos o estudios de rendimiento de cultivos.
lower_bound_simul = 326.4532
upper_bound_simul = 390.8709

# 3. Itera sobre cada fila del DataFrame y realiza la simulación para 'produc_simul_kg_haMC'.
for index, row in df.iterrows():
    # a. Genera una lista de números aleatorios distribuidos uniformemente.
    simulated_values_simul = np.random.uniform(lower_bound_simul, upper_bound_simul, num_iterations_simul)

    # b. Calcula la media de estos valores simulados.
    mean_simulated_value_simul = np.mean(simulated_values_simul)

    # c. Asigna esta media a la nueva columna 'produc_simul_kg_haMC' de la fila actual.
    df.loc[index, 'produc_simul_kg_haMC'] = mean_simulated_value_simul

# --- Calcular 'prod_est_kg_ha' ---
# Objetivo: Calcular la producción estimada total en kg por hectárea para cada parcela.
# Esta es una métrica clave que integra el área de la parcela, el factor de productividad
# y el factor de arrime, proporcionando una estimación holística de la producción.
# La fórmula integra los resultados de las simulaciones previas con los datos de área.
df['prod_est_kg_ha'] = df['Area (ha)'] * df['fact_prod_ML'] * df['fact_arrim']

print("Simulaciones y cálculos intermedios completados. Mostrando las primeras 10 filas del DataFrame:")
# Muestra las primeras 10 filas del DataFrame para verificar que todas las columnas simuladas
# y calculadas se han añadido y actualizado correctamente, y para una revisión inicial de los valores.
df.head(10)
```
